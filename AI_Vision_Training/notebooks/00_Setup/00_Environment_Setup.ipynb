## Part 1　基本の言葉 / Basic vocabulary

#### 日本語：Python とは何か
Python は「コンピュータに仕事をさせるための言葉（プログラミング言語）」の 1 つです。

たとえば、次のような仕事を Python に頼めます。

- 計算する（100.3 が 100±0.5 の範囲に入っているか判定する）
- 画像を読み込む（カメラで撮ったシートの写真）
- 画像の中の黒い点を探す
- 結果を表やグラフにする

Python の長所は **読んで意味が分かりやすい** ことです。次の 1 行を見てください。

```python
width = 100.3
```

「width（幅）は 100.3」と、ほぼ英語のまま読めます。あなたが書いた Python の文は、
**1 行ずつ上から順に** 実行されます。
#### English: What is Python
Python is one of the languages we use to tell a computer what to do.

You can ask Python to:

- calculate (is 100.3 inside the range 100 ± 0.5?)
- read an image (a photo of the sheet taken by a camera)
- look for dark spots inside that image
- turn the result into a table or a graph

Python's big advantage is that it **reads almost like English**:

```python
width = 100.3
```

Your Python statements are executed **one line at a time, from top to bottom**.

#### 日本語：ターミナル（PowerShell）とは何か
ターミナルは「文字でコンピュータに命令する窓」です。Windows では **PowerShell** を使います。
マウスでクリックする代わりに、文字で命令します。

**PowerShell の開き方（3 通り）**

1. 画面下の検索欄に `PowerShell` と入力して開く。
2. スタートボタンを右クリック →「ターミナル」または「Windows PowerShell」。
3. エクスプローラーでフォルダを開き、アドレス欄に `powershell` と入力して Enter。

開くと、次のような行が表示されます。これを **プロンプト** と呼びます。

```text
PS C:\Users\<あなたの名前>>
```

`PS` は PowerShell が待機中という意味、その後ろが「今いるフォルダ」です。
#### English: What is a terminal (PowerShell)
A terminal is a window where you give the computer instructions as text. On Windows we use
**PowerShell**: instead of clicking with the mouse, you type commands.

**Three ways to open PowerShell**

1. Type `PowerShell` in the Windows search box and open it.
2. Right-click the Start button -> "Terminal" / "Windows PowerShell".
3. Open a folder in File Explorer, type `powershell` in the address bar, press Enter.

You will see a line called the **prompt**:

```text
PS C:\Users\<your-name>>
```

`PS` means PowerShell is waiting; the rest is the folder you are currently in.

#### 日本語：仮想環境（virtual environment）とは何か、なぜ必要か
**仮想環境** とは「この研修専用の、Python の道具箱」です。

パソコン全体で 1 つの道具箱を共有すると、次の問題が起きます。

| 問題 | 具体例 |
|---|---|
| バージョン衝突 | 別の仕事で古い OpenCV が必要になり、研修の OpenCV が壊れる |
| 再現できない | 「私の PC では動くのに」が起きる |
| 壊れたときの復旧が大変 | どのパッケージが原因か分からない |

仮想環境を作ると、**そのフォルダの中だけ** にパッケージが入ります。
他の仕事や他の Python には一切影響しません。壊れたらフォルダを消して作り直せます。

> **日本語：** この研修では 2 つの仮想環境を使います。
> `aivision`（基本＋YOLO）と `aivision-anomalib`（PatchCore 用）です。
> 理由は Day 5 のノートブックで説明します。

**English:** A virtual environment is a private toolbox of Python packages for one project.

Sharing one global toolbox causes version conflicts, "works on my machine" problems, and painful
repair. Inside a virtual environment, packages are installed **only in that folder** and cannot
damage anything else. If it breaks, delete the folder and build it again.

This course uses two environments: `aivision` (basics + YOLO) and `aivision-anomalib`
(PatchCore). Day 5 explains exactly why they are separate.
#### English: What a virtual environment is, and why it matters

#### 日本語：Jupyter ノートブック・セル・カーネル
**Jupyter ノートブック** は「説明」と「動くプログラム」と「結果」を 1 つのファイルに
まとめた文書です。ファイルの拡張子は `.ipynb` です。

- **セル（cell）**：ノートブックの中の 1 つの箱。説明用（Markdown）とプログラム用（コード）があります。
- **カーネル（kernel）**：コードセルを実際に実行する「Python の本体」。
  ノートブックは「どのカーネルで動かすか」を覚えています。
- **変数はカーネルの中に残ります**。セル A で作った変数は、セル B でも使えます。
  ただし **カーネルを再起動すると、変数は全部消えます**。

> **日本語：** 「カーネルを間違える」は、この研修で最も多いトラブルの 1 つです。
> 正しいカーネルかどうかは、次のセルの `sys.executable` で確認できます。

**English:** A Jupyter notebook is a document that combines explanations, runnable programs and
results in one `.ipynb` file.

- A **cell** is one box inside the notebook (markdown for text, code for programs).
- A **kernel** is the actual Python process that runs the code cells. The notebook remembers
  which kernel it uses.
- **Variables live inside the kernel.** A variable created in cell A is still there in cell B,
  but **restarting the kernel deletes every variable**.

Choosing the wrong kernel is one of the most common problems in this course. The next code cell
shows `sys.executable`, which tells you exactly which Python is running.
#### English: Jupyter notebook, cell and kernel

## Part 3　ノートブックから入れる方法（%pip）/ Installing from inside the notebook

#### 日本語：`%pip` とは何か
`%` で始まる行は **マジックコマンド** と呼ばれる、Jupyter 専用の命令です。
`%pip` は「このノートブックが使っているカーネルに対して pip を実行する」という意味です。

```python
%pip install requests
```

**日本語：** これは Python の文法ではありません。だから PowerShell に書いても動きません。
逆に、`python -m pip install ...` をノートブックに書いても動きません。

> **この研修の方針：** インストールは **最初に 1 回だけ**。毎回のノートブック実行では
> インストールしません。理由は 3 つあります。
> 1. 毎回数分待たされる
> 2. ネットワークが無い日に動かなくなる
> 3. 環境が少しずつ変わって、原因が分からなくなる

**English:** A line starting with `%` is a **magic command** — a Jupyter-only instruction.
`%pip` runs pip against the kernel that this notebook is using.

This is not Python syntax: it does not work in PowerShell, and `python -m pip install ...`
does not work inside a notebook.

**Policy of this course:** install **once**, at setup time. Routine notebooks never install
packages, because (1) it wastes minutes, (2) it fails without network, and (3) it silently
changes the environment.
#### English: What `%pip` is

**日本語：** 下のセルは **実行しないでください**。書き方の見本です。必要なときだけ、
`#` を消して実行します。

**English:** Do **not** run the next cell. It is a template; remove the `#` only if you
really need to install something.

#### 日本語：図の日本語フォントを設定する
**日本語：** Matplotlib の既定フォントは日本語を表示できません。そのままだと図のタイトルが `□□□` になります。次のセルで、日本語を表示できるフォントを探して設定します。Mac や Linux では自動で別のフォントが選ばれます。
#### English: Configure Japanese text in figures
**English:** Matplotlib's default font cannot draw Japanese characters, so figure titles would appear as `□□□`. The next cell looks for a Japanese font and selects it. On macOS or Linux a different font will be chosen automatically.

````python
import platform
from pathlib import Path

from matplotlib import font_manager, rcParams

def find_japanese_font():
    """日本語を表示できるフォントを探す。見つからなければ None を返す。
    Locate a font that can render Japanese. Returns None when nothing is found."""
    system = platform.system()
    if system == "Windows":
        folders = [Path("C:/Windows/Fonts")]
    elif system == "Darwin":
        folders = [Path("/System/Library/Fonts"), Path("/Library/Fonts"), Path.home() / "Library/Fonts"]
    else:
        folders = [Path("/usr/share/fonts"), Path("/usr/local/share/fonts"), Path.home() / ".fonts"]

    preferred = [
        "NotoSansJP-Regular", "NotoSansCJK-Regular", "YuGothR", "meiryo", "msgothic",
        "BIZ-UDGothicR", "NotoSansJP", "HiraginoSans",
    ]
    found = []
    for folder in folders:
        if not folder.is_dir():
            continue
        for path in folder.rglob("*"):
            if path.suffix.lower() not in {".ttf", ".ttc", ".otf"}:
                continue
            name = path.stem
            if any(key.lower() in name.lower() for key in preferred):
                found.append(path)

    def rank(path):
        name = path.stem.lower()
        return (0 if "regular" in name or name.endswith("r") else 1, len(name))

    return sorted(found, key=rank)[0] if found else None


japanese_font = find_japanese_font()
if japanese_font is None:
    print("日本語フォントが見つかりません / no Japanese font found.")
    print("図のタイトルが □ になる場合は、英字のタイトルを使ってください。")
    print("If figure titles show boxes, use English titles instead.")
else:
    font_manager.fontManager.addfont(str(japanese_font))
    family_name = font_manager.FontProperties(fname=str(japanese_font)).get_name()
    rcParams["font.family"] = [family_name, "DejaVu Sans"]
    rcParams["axes.unicode_minus"] = False
    print("図の日本語フォント / figure font:", family_name)
    print("  ->", japanese_font)
````

#### 日本語：数字の集まりが画像になる
画像は魔法ではありません。**数字の表** です。

- 1 枚の画像は、小さな四角（**ピクセル**）が縦横に並んだものです。
- 各ピクセルは 1 つの数値を持ちます。0 が黒、255 が白です。

**日本語：** つまり、Python で数字の表を作れば、それがそのまま画像になります。

#### English: An image is just a grid of numbers

````python
import numpy as np

# 8 x 8 の「ほぼ白」の画像を作る / build an 8x8 nearly-white image
image = np.full((8, 8), 230, dtype=np.uint8)
print("shape (height, width):", image.shape)
print("data type            :", image.dtype)
print()
print(image)
````

## Part 7　環境準備レポート / Environment readiness report

#### 日本語：レポートを自動で作る
**日本語：** 最後に、今の環境を 1 枚のレポートにまとめます。
`submissions/reports/environment_report.json` に保存されます。
提出するときに、この 2 つのセルの出力をそのまま貼り付けてください。

**English:** Finally, summarise the environment into a report written to
`submissions/reports/environment_report.json`. Paste the output of this cell into your
submission.
#### English: Generate the report automatically
The report records the interpreter, the package versions and whether each package could be
imported. If a package is missing the report says so — nothing is hidden.

````python
import json
import platform
import sys
from importlib import metadata
from pathlib import Path

PACKAGES = [
    ("numpy", "numpy"),
    ("matplotlib", "matplotlib"),
    ("PIL", "pillow"),
    ("pandas", "pandas"),
    ("cv2", "opencv-python"),
    ("skimage", "scikit-image"),
]

report = {
    "python": {
        "version": platform.python_version(),
        "executable": sys.executable,
        "is_virtualenv": sys.prefix != getattr(sys, "base_prefix", sys.prefix),
    },
    "platform": f"{platform.system()} {platform.release()} ({platform.machine()})",
    "packages": {},
    "assets_ready": all(
        (ASSETS / name).exists() for name in ("diagrams", "normal_samples", "defect_samples")
    ),
}

for import_name, dist_name in PACKAGES:
    try:
        __import__(import_name)
        installed = True
    except ImportError:
        installed = False
    try:
        version = metadata.version(dist_name)
    except metadata.PackageNotFoundError:
        version = None
    report["packages"][dist_name] = {"importable": installed, "version": version}

report_path = COURSE_ROOT / "submissions" / "reports" / "environment_report.json"
report_path.parent.mkdir(parents=True, exist_ok=True)
report_path.write_text(json.dumps(report, indent=2, ensure_ascii=False), encoding="utf-8")

print(json.dumps(report, indent=2, ensure_ascii=False))
print()
print("保存しました / saved to:", report_path)

missing = [name for name, info in report["packages"].items() if not info["importable"]]
if missing:
    print("未導入 / missing:", ", ".join(missing))
    print("=> Part 2 手順 4 を実行してください / run Part 2 Step 4.")
else:
    print("READY / 準備完了。第1日のノートブックへ進んでください。")
    print("The environment is ready. Continue with Day 1 Notebook 01.")
````
