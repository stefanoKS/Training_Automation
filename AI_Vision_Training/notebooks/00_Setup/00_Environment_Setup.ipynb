#### 最初に知っておくこと：コマンドと Python コードは別物です

これから 2 種類の「命令」が出てきます。**混ぜると必ず失敗します。**

**1. CMD コマンド（コマンド プロンプトに書くもの）**

```cmd
python --version
```

これは Windows のコマンド プロンプトに書きます。`python` という **道具** を呼び出しています。

**2. Python コード（このノートブックのコードセルに書くもの）**

```python
print("こんにちは")
```

これは Python という **言語** で書かれた文です。CMD に直接書くとエラーになります。

> ノートブックのコードセルに `python -m pip install ...` のような
> CMD コマンドを書いてはいけません。Python の構文エラーになります。

#### First, know the difference: commands and Python code are not the same
There are two different kinds of instruction in this course, and mixing them always fails:

1. **CMD commands** — typed into Command Prompt, e.g. `python --version`.
2. **Python code** — typed into a notebook code cell, e.g. `print("hello")`.

Never paste a CMD command into a normal Python cell. Use CMD for the official installation; Part 3's `%pip` is a Jupyter-only exception.

#### Python とは何か

<!-- ne:begin after-heading -->
**Python भनेको के हो**
<!-- ne-pair: What is Python -->
<!-- ne:end -->
Python は「コンピュータに仕事をさせるための言葉（プログラミング言語）」の 1 つです。

たとえば、次のような仕事を Python に頼めます。

- 計算する（100.3 が 100±0.5 の範囲に入っているか判定する）
- 画像を読み込む（カメラで撮ったシートの写真）
- 画像の中の黒い点を探す
- 結果を表やグラフにする

Python の長所は **読んで意味が分かりやすい** ことです。次の 1 行を見てください。

```python
width = 100.3
```

「width（幅）は 100.3」と、ほぼ英語のまま読めます。あなたが書いた Python の文は、
**1 行ずつ上から順に** 実行されます。
#### What is Python
Python is one of the languages we use to tell a computer what to do.

You can ask Python to:

- calculate (is 100.3 inside the range 100 ± 0.5?)
- read an image (a photo of the sheet taken by a camera)
- look for dark spots inside that image
- turn the result into a table or a graph

Python's big advantage is that it **reads almost like English**:

```python
width = 100.3
```

Your Python statements are executed **one line at a time, from top to bottom**.

<!-- ne:begin end -->
Python "कम्प्युटरलाई काम लगाउने भाषा (प्रोग्रामिङ भाषा)" मध्ये एक हो।

- चित्र पढ्ने (क्यामेराले खिचेको सिटको फोटो)
- चित्रभित्रको कालो थोप्लो खोज्ने

#### コマンド プロンプト（CMD）とは何か

#### What a virtual environment is, and why it matters
A virtual environment is a private toolbox of Python packages for one project.

Sharing one global toolbox causes version conflicts, "works on my machine" problems, and painful
repair. Inside a virtual environment, packages are installed **only in that folder** and cannot
damage anything else. If it breaks, delete the folder and build it again.

This course uses two environments: `aivision` (basics + YOLO) and `aivision-anomalib`
(PatchCore). Day 5 explains exactly why they are separate.

#### Jupyter notebook, cell and kernel
A Jupyter notebook is a document that combines explanations, runnable programs and
results in one `.ipynb` file.

- A **cell** is one box inside the notebook (markdown for text, code for programs).
- A **kernel** is the actual Python process that runs the code cells. The notebook remembers
  which kernel it uses.
- **Variables live inside the kernel.** A variable created in cell A is still there in cell B,
  but **restarting the kernel deletes every variable**.

Choosing the wrong kernel is one of the most common problems in this course. The next code cell
shows `sys.executable`, which tells you exactly which Python is running.

**Jupyter notebook** भनेको "व्याख्या", "चल्ने प्रोग्राम" र "नतिजा" एउटै फाइलमा राखिएको कागजात हो।

- **kernel**: कोड सेल वास्तवमा चलाउने "Python को मुख्य भाग"।
  नोटबुकले "कुन kernel मा चलाउने" सम्झिरहन्छ।
- **variable kernel भित्र रहन्छ**। सेल A मा बनाइएको variable सेल B मा पनि प्रयोग गर्न सकिन्छ।

> "kernel गलत छान्ने" यो तालिमको सबैभन्दा सामान्य समस्यामध्ये एक हो।
> ठीक kernel हो कि होइन अर्को सेलको `sys.executable` ले जाँच्न सकिन्छ।

#### Calculating and variables
Python also works as a calculator. `=` means "store the value on the right under this name".
It does **not** mean "is equal to"; think of it as putting a value into a named box.

<!-- ne:begin end -->
Python क्याल्कुलेटरका रूपमा पनि प्रयोग गर्न सकिन्छ। `=` भनेको "दायाँपट्टिको नतिजालाई नाम दिने" हो।

````python
# ▸ この値を変更 / Change this value：値を変えて、結果を予想してから実行します。
# ▸ Change this value: edit it below, predict the result, then run.
# シートの幅を測った値 / a measured sheet width
width_mm = 100.3
nominal_mm = 100.0
tolerance_mm = 0.5

difference = width_mm - nominal_mm
limit = tolerance_mm

print("measured width :", width_mm, "mm")
print("nominal width  :", nominal_mm, "mm")
print("difference     :", round(difference, 3), "mm")
print("limit          :", limit, "mm")
print()
print("OK?" , abs(difference) <= limit)
````

#### 数字の集まりが画像になる

画像は魔法ではありません。**数字の表** です。

#### An image is just a grid of numbers
An image is not magic — it is a table of numbers. Each **pixel** holds a value
(0 = black, 255 = white); a colour image holds three values (red, green, blue).
Create the table of numbers in Python and you have created an image.

<!-- ne:begin end -->
चित्र जादु होइन। **सङ्ख्याको तालिका** हो।

- एउटा चित्र साना चौकोण (**pixel**) तलमाथि र छेउछाउ मिलेर राखिएको हुन्छ।
- प्रत्येक pixel मा एउटा सङ्ख्या हुन्छ। 0 कालो, 255 सेतो।
- रङ चित्रमा रातो, हरियो, निलो गरी ३ सङ्ख्या हुन्छन्।

अर्थात् Python मा सङ्ख्याको तालिका बनायो भने त्यही चित्र बन्छ।
अर्को सेलमा वास्तवमा बनाइन्छ।
<!-- ne:end -->


````python
# ▸ 実行のみ / Run only：このまま実行します。書き換えません。
# ▸ Run only: press Shift + Enter and leave the code as it is.
import numpy as np

# 8 x 8 の「ほぼ白」の画像を作る / build an 8x8 nearly-white image
image = np.full((8, 8), 230, dtype=np.uint8)
print("shape (height, width):", image.shape)
print("data type            :", image.dtype)
print()
print(image)
````

````python
# ▸ 実行のみ / Run only：このまま実行します。書き換えません。
# ▸ Run only: press Shift + Enter and leave the code as it is.
# 一部のピクセルを黒くする = 欠陥を作る / darken some pixels: that is a defect
image[2:4, 5:7] = 20

print("3 行目の値 / values in row 3:", image[3])
print("変わったピクセル数 / changed pixels:", int((image < 100).sum()))
````

#### Displaying with Matplotlib
Inside Jupyter, never use `cv2.imshow()`: it opens a separate window and can freeze the notebook.
Use **Matplotlib** instead — the whole course does it this way.

`plt.imshow()` prepares the picture, `plt.show()` renders it, and `interpolation="nearest"` keeps

The picture shown here is a **synthetic** 8 x 8 image, not a factory photograph.

`interpolation="nearest"` ले "ठुलो बनाउँदा पनि सीमा धमिलो नबनाउने" निर्देश दिन्छ।
