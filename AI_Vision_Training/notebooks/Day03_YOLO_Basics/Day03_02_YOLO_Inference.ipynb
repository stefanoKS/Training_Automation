````python
# ▸ 実行のみ / Run only：このまま実行します。書き換えません。
# ▸ Run only: press Shift + Enter and leave the code as it is.
import cv2
import numpy as np
import pandas as pd
from PIL import Image
import matplotlib.pyplot as plt

def show(images, titles=None, figsize=(14, 5)):
    """画像を並べて表示する / display images side by side."""
    if not isinstance(images, (list, tuple)):
        images = [images]
    if titles is None:
        titles = [""] * len(images)
    fig, axes = plt.subplots(1, len(images), figsize=figsize)
    if len(images) == 1:
        axes = [axes]
    for ax, image, title in zip(axes, images, titles):
        ax.imshow(image)
        ax.set_title(title, fontsize=10)
        ax.set_xticks([]); ax.set_yticks([])
    plt.tight_layout()
    plt.show()


def detections_to_frame(result):
    """検出結果を pandas の表にする / convert detections into a DataFrame.

    ``result.boxes.data`` の 6 列は
    (x1, y1, x2, y2, confidence, class_id) です。
    """
    columns = ["class_id", "name", "confidence", "x1", "y1", "x2", "y2"]
    boxes = result.boxes
    if len(boxes) == 0:
        return pd.DataFrame(columns=columns)

    data = boxes.data.cpu().numpy()
    return pd.DataFrame({
        "class_id": data[:, 5].astype(int),
        "name": [result.names[int(class_id)] for class_id in data[:, 5]],
        "confidence": data[:, 4].round(4),
        "x1": data[:, 0].round(2),
        "y1": data[:, 1].round(2),
        "x2": data[:, 2].round(2),
        "y2": data[:, 3].round(2),
    })
````

## Part 2　モデルを読み込む / Loading the model

#### 初回だけダウンロードする
モデルのファイル（約 5 MB）は、**初回だけ** ダウンロードされます。
2 回目以降は `assets/models/` から読み込むので、ネットワークが無くても動きます。

**ダウンロードに失敗したときの手順**

   `yolo26n.pt` を手動でダウンロードして `assets/models/` に置く
3. 社内プロキシがある場合は、管理者に確認する

> エラーはそのまま記録して、講師に相談してください。

#### Downloaded once, cached afterwards
The model file (~5 MB) is downloaded **only once**. After that it is loaded from
`assets/models/`, so no network is needed.

**If the download fails**

1. run the cell again (transient failures are common)
2. open `https://github.com/ultralytics/assets/releases` in a browser, download `yolo26n.pt`
   manually and place it in `assets/models/`
3. if the company uses a proxy, ask IT

**Never invent a plausible result when the download fails.** Record the error and ask the instructor.

````python
# ▸ 実行のみ / Run only：このまま実行します。書き換えません。
# ▸ Run only: press Shift + Enter and leave the code as it is.
import shutil
from pathlib import Path

MODELS_DIR = ASSETS / "models"
MODELS_DIR.mkdir(parents=True, exist_ok=True)

def load_pretrained_yolo(model_name="yolo26n.pt"):
    """学習済みモデルを読み込む。初回だけダウンロードする。
    Load a pretrained model, downloading it only on the first run."""
    from ultralytics import YOLO

    weights_path = MODELS_DIR / model_name
    if not weights_path.is_file():
        print(f"初回のみダウンロードします / downloading {model_name} (first run only)...")
        print("  ネットワークに接続できない場合は、次を確認してください:")
        print("  - 社内プロキシの設定 / corporate proxy settings")
        print("  - もう一度このセルを実行する / run this cell again")
        model = YOLO(model_name)          # ダウンロードはカレントフォルダに行われる
        downloaded = Path(model_name)
        if downloaded.is_file():
            shutil.move(str(downloaded), str(weights_path))
            print("保存しました / cached to:", weights_path)
        else:
            raise RuntimeError(
                f"モデルを取得できませんでした / could not obtain {model_name}. "
                "ネットワークを確認して再実行してください / check the network and retry."
            )
    return YOLO(str(weights_path))
````

#### 画像を読み込んで予測する
`model.predict()` の主な引数です。

| 引数 | 意味 | 例 |
|---|---|---|
| `source` | 画像のパス | `"image.jpg"` |
| `conf` | これ未満の検出は**推論のときに捨てる** | `0.25` |

戻り値は「結果のリスト」です。1 枚の画像なら `results[0]` に 1 枚分が入っています。

このあと Part 1 のように「低いしきい値で 1 回だけ推論して、あとから絞り込む」使い方と、
`conf` を毎回変えて推論し直す使い方は別物です。第 3 日のノートブック 03 で詳しく見ます。

#### Load an image and predict
The main arguments of `model.predict()`:

| Argument | Meaning | Example |
|---|---|---|
| `source` | image path | `"image.jpg"` |
| `conf` | **discard at inference time** below this value | `0.25` |
| `imgsz` | inference size (smaller is faster) | `640` |
| `verbose` | print logs | `False` |

The return value is a **list of results**; for a single image, `results[0]` holds that image.

Candidates discarded by `conf` do not come back if you lower the threshold afterwards.
"Run once at a low threshold and filter afterwards" (as in Part 1 of Notebook 03) is a different
pattern from re-running the model with a new `conf` each time.


````python
# ▸ この値を変更 / Change this value：値を変えて、結果を予想してから実行します。
# ▸ Change this value: edit it below, predict the result, then run.
import ultralytics

# ultralytics パッケージに同梱されている公式サンプル画像 / the sample image shipped with the package
sample_image = Path(ultralytics.__file__).parent / "assets" / "bus.jpg"
print("画像 / image:", sample_image)
print("存在 / exists:", sample_image.is_file())

if not sample_image.is_file():
    raise FileNotFoundError(
        "同梱のサンプル画像が見つかりません。ultralytics を再インストールしてください。"
        " / the bundled sample image is missing; reinstall ultralytics."
    )

results = model.predict(source=str(sample_image), conf=0.25, imgsz=640, verbose=False)
result = results[0]

print()
print("検出の数 / number of detections:", len(result.boxes))
print("画像サイズ / image shape:", result.orig_shape)
````

````python
# ▸ 実行のみ / Run only：このまま実行します。書き換えません。
# ▸ Run only: press Shift + Enter and leave the code as it is.
annotated_bgr = result.plot()                     # 描画した BGR 画像 / drawn BGR image
annotated_rgb = cv2.cvtColor(annotated_bgr, cv2.COLOR_BGR2RGB)

show([cv2.cvtColor(cv2.imread(str(sample_image)), cv2.COLOR_BGR2RGB), annotated_rgb],
     ["input / 入力画像", "YOLO prediction / 予測結果"], figsize=(14, 5))
````

#### 何が見えたか
画像に **バス 1 台と、数人の人** が写っています。
それぞれに四角が描かれ、クラス名と confidence が表示されます。
この研修で記録した実行では `bus 0.88` `person 0.87` のように表示されました。
値はモデル・画像・`conf` で変わります。

ここで大切なのは、**モデルが学習したクラスを返せていることの確認** です。
PVC の欠陥検出は、まだできません。

**検出が 0 件だったときは？** ここでいう 0 件は
**「設定したクラス（COCO の 80 種類）で残った検出が 0 件」** という意味です。
「この画像に欠陥が無い」証拠にはなりません。

0 件にはいくつもの理由がありえます。

- この画像に COCO のクラスが本当に写っていない
- `conf` のしきい値で、出ていた候補がすべて捨てられた
- 推論そのものが失敗した（パスの誤り、画像が読めない、など）

逆に、しきい値を下げると **COCO のクラスとして誤った検出が出ることもあります**
（例：背景の模様を `stop sign` と表示する）。COCO 事前学習モデルは、
PVC の欠陥について意味のある判定をするようには作られていません。
だから 0 件を「OK」と読み替えず、**何を学習したモデルか、どのしきい値で実行したか、

#### What you should see
The image contains **one bus and several people**. Each has a box labelled with a class name and
a confidence. In this course's recorded run the labels read like `bus 0.88` and `person 0.87`;
the values change with the model, the image and `conf`.

What matters here is confirming that **the model returns the classes it learned**.
It still cannot detect PVC defects.

**What if there are zero detections?** Here zero means
**"no retained detections of the configured classes (the 80 COCO classes)"**.
It is not evidence that the image contains no defect.

Zero can have several causes:

- the image really contains none of the COCO classes
- the `conf` threshold discarded every candidate
- inference itself failed (wrong path, unreadable image, and so on)

Conversely, lowering the threshold can produce **spurious COCO-class detections**
(for example a background pattern labelled `stop sign`). The COCO-pretrained model is not designed
to give a meaningful judgement about PVC defects. Never read zero as OK: read it together with
**what the model was trained on, which threshold you ran, and what the image shows**.

## Part 4　結果を取り出す / Extracting the results

The six columns of `boxes.data` are, left to right,

#### `utf-8-sig` を付ける理由
Excel で CSV を開くと、`utf-8` のままだと日本語が文字化けすることがあります。
`utf-8-sig` を付けると、Excel でも正しく表示されます。

#### Why `utf-8-sig`
Plain `utf-8` CSVs often show garbled Japanese in Excel; `utf-8-sig` adds a byte
order mark so Excel reads it correctly.
