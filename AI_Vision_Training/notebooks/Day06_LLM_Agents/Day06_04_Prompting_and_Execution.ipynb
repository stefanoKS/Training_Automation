# 第6日 03：コーディングエージェントへの伝え方 / Day 6 03: Communicating with coding agents

**目安：80分 / Estimated duration: 80 min.** [第6日時間割](../DAY06_SYLLABUS_ja_en.md)を参照。/ See the [sole Day 6 timetable](../DAY06_SYLLABUS_ja_en.md).

**目標 / Objectives:** 仕様から指示案を作り、実装前の計画、許可、段階的な作業と具体的な修正依頼を行う。/ Draft agent instructions from approved requirements; request a plan and approval; stage changes and correct observable errors.



## 1. 依頼の四段階 / Four stages of requests

探索（exploration）は問題と選択肢を調べる段階、仕様（specification）は何を満たすか決める段階、計画（planning）はどう作るか調べる段階、実装（implementation）は承認された変更を行う段階です。曖昧なアイデアから直接実装すると、AIが大切な条件を推測してしまいます。

Exploration investigates problems and options; specification defines expected behavior; planning investigates how to build it; implementation makes approved changes. Jumping straight from a vague idea to implementation encourages unchecked assumptions.

![人、ChatGPT、コーディングエージェントが受け渡す情報 / Human review between specification and coding agent](../assets/diagrams/D06-09.svg)

**図 D06-09 / Figure D06-09:** 人が仕様と計画を確認します。/ The human reviews the specification and plan. **問い / Reflect:** ChatGPTが書いた指示はそのまま実行してよいですか？ / Should a ChatGPT draft be run without review?



## 2. 指示を組み立てる / Build the instruction

次の9点を記入します。役割：既存の作業を尊重する作成者。目的：記録業務を楽にする。背景：検査担当者が使う。要件：承認済みの入力、保存、絞り込み、CSV。制約：架空のローカルデータだけ、他の教材や本番設備に触れない。進め方：調査→計画→承認→作成→試験→報告。受入基準：例を使って確認できる挙動。成果物：アプリ、使い方、試験記録。報告：変更箇所、試験、残る問題。

Fill nine slots: role (respect existing work), objective (make records easier), context (inspection operators), requirements (approved fields, persistence, filters, CSV), constraints (fictional local data only; no other training files or live systems), procedure (inspect → plan → approval → build → test → report), acceptance checks (observable example outcomes), deliverables (app, instructions, test record), and report (changes, tests and remaining issues). These are aids for clarity, not magic words.

![曖昧な指示と構造化した指示 / Poor and improved instructions](../assets/diagrams/D06-06.svg)

**図 D06-06 / Figure D06-06:** 「何を作るか」と「触れてよい範囲」の両方を書く。/ State both desired behavior and permitted scope. **問い / Reflect:** 制約を省くと何が起こりますか？ / What could happen if constraints are omitted?



## 3. 計画だけを先に依頼する / Ask for a plan first

「リポジトリを調べ、既存機能、変更予定ファイル、使える手段、危険、試験方法を示してください。**まだファイルを変更しないでください。**」と伝えます。例えばAIが画像自動判定やクラウドDBを提案しても、この学習用アプリには不要です。「それらを除き、ローカルの記録機能だけで計画し直してください」と返します。

Say: “Inspect the repository. List existing behavior, planned file changes, available tools, risks and tests. **Do not modify files yet.**” If the agent proposes automatic image inspection or a cloud database, reject them: “Remove those additions. Re-plan only the approved local record workflow.”

![依頼から確認・修正・最終確認までの架空のやりとり / Fictional request-to-review timeline with plan rejection](../assets/diagrams/D06-10.svg)

**図 D06-10 / Figure D06-10:** 不要な案は実装前に戻す。/ Reject unnecessary features before implementation. **問い / Reflect:** 計画が範囲外なら次に何と言いますか？ / What do you say if the plan exceeds scope?



## 4. 作業を小さくする / Keep stages small

段階1：画面案。段階2：入力確認。段階3：保存と再起動。段階4：絞り込みとCSV。段階5：試験と説明。各段階で「次へ進んでよいか」を確認するチェックゲートを置きます。要件承認、計画確認、範囲内の変更、試験、人による受入のいずれかが失敗したら、該当段階へ戻ります。

Stages: (1) UI concept, (2) validation, (3) persistence, (4) filters and CSV, (5) tests and documentation. At each checkgate ask whether to proceed. If requirements approval, plan review, scope review, tests or human acceptance fails, return to the relevant stage.

![承認、範囲、試験と受入のチェックゲート / Quality gates with return paths](../assets/diagrams/D06-11.svg)

**図 D06-11 / Figure D06-11:** 「試験済み」と「受入済み」は別です。/ “Tested” differs from “accepted.” **問い / Reflect:** 試験に通っても人が確認する理由は？ / Why check manually even after tests pass?



## 5. 修正は観察から / Correct using observations

悪い例：「アプリが変。直して」。良い例：「ロット番号を空にして保存を押すと、記録が増えます（観察）。保存せず日本語のエラーを出すはずです（期待）。空欄で保存を押して再現し、修正してください。既存の保存機能と他の教材を変えず、有効なロットも引き続き保存できることを確認してください（範囲と検証）。」

Weak: “The app is wrong. Fix it.” Better: “Saving with an empty lot adds a row (observed). It should refuse the save and show a Japanese error (expected). Reproduce by clearing the lot and pressing Save. Preserve valid saves and unrelated lessons; demonstrate both failing and valid cases after the fix (scope and verification).”

この形は「観察→期待→再現方法→制約→確認」です。ボタンがない、ラベルが英語、CSV列が欠ける、再起動後に消える、絞り込み件数が合わない、といった場合も同じ形で伝えます。CSVがないなら、「承認した仕様にCSVがあるのにボタンがありません。既存の保存・一覧を維持し、列順と架空の2件を確認してください」と依頼します。

This is observed → expected → reproduction → constraints → verification. Use it for missing buttons, untranslated labels, wrong CSV columns, lost records or incorrect counts. For absent CSV: “The approved specification requires CSV, but the button is absent. Add only export, preserve save/list, and check the agreed column order and two fictional rows.”

**問い / Reflect:** 「CSVがおかしい」だけでは何が足りませんか？ / What is missing from “CSV is broken”?



## 6. 長い仕事を引き継ぐ / Maintain context across long tasks

会話が長くなると重要な条件が見えにくくなります。承認した仕様、作業の進み具合、変更したファイル、試験結果、次の作業を文書に残します。別のエージェントには、まず既存の状態を調べさせてから続けさせます。チェックポイントは後戻り・引き継ぎの目印であり、完了の証明ではありません。

In long conversations, important requirements can be missed. Save the approved specification, progress, changed files, test results and next steps. A receiving agent must inspect existing state before continuing. A checkpoint helps resumption but does not prove correctness.

![仕様・進捗・試験・課題を次の担当へ渡す / Handover package to receiving agent](../assets/diagrams/D06-13.svg)

**図 D06-13 / Figure D06-13:** 「どこまで確認したか」を残す。/ Record what has actually been checked. **問い / Reflect:** 次のエージェントに渡すものを三つ挙げてください。/ Name three handover items.



## 7. 演習（合計20分）/ Practice (20 min total)

**D / Activity D (5 min):** 「いいアプリを作って」を、利用者、入力、出力、制約、試験を含む指示へ直す。例：[指示例](../examples/example_agent_prompts_ja_en.md)。**出力 / Output:** 改善した指示。**間違い / Pitfall:** 「必ず完璧に」だけ足す。/ Rewrite “Make a good app” with users, inputs, outputs, scope and checks. Pitfall: adding only “make it perfect.”

**E / Activity E (4 min):** 「不良記録アプリを作って。全部のファイルを自由に直して」という架空の依頼から弱点を三つ探す。**出力 / Output:** 3点と修正案。**間違い / Pitfall:** 範囲外の変更を許す。/ Find three omissions or unsafe permissions in “Make the app; change any file you like.” Output: three problems and corrections. Pitfall: broad permissions.

**F / Activity F (4 min):** 「クラウドDBを追加し、本番検査データを読み込む」という架空計画を却下し、計画のみを求め直す。**出力 / Output:** 修正依頼。**間違い / Pitfall:** 実装してから確認する。/ Reject the fictional cloud/production-data plan; ask for a revised plan only. Pitfall: checking after implementation.

**修正演習 / Correction activity (7 min):** CSVボタンがない架空の結果について、観察→期待→再現→制約→検証を書きます。**出力 / Output:** 短い修正指示。**間違い / Pitfall:** 全アプリを作り直す。/ Write a focused correction for absent CSV. Pitfall: requesting a complete rewrite.

**完了チェック / Completion:** 実装より先に計画を求め、範囲外の提案を止め、観察に基づく修正指示を書けた。/ Ask for a plan first, reject out-of-scope work and write an evidence-based correction.
